```python
import sklearn as sk
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import KFold
import pandas as pd
import numpy as np

df = pd.read_csv("iris/iris.data",header=None)

X = df.drop([4],axis=1)
y = df[4]
```

```python
K = []
scores = {}
bestScore = 0
bestK = 0

for k in range(3,31,2):
    model = sk.neighbors.KNeighborsClassifier(n_neighbors=k,metric="cosine")
    model.fit(X, y)

    score = model.score(X,y)

    if (score > bestScore):
        bestScore = score
        bestK = k

    scores[k] = [k, score]
```

```python
model = sk.neighbors.KNeighborsClassifier(n_neighbors=bestK)
model.fit(X, y)
y_pred = model.predict(X)
print("K-Neighbors  Score")
for key , value in scores.items():
    print(f"[{key} : {round(value[1],3)}]")
```

```python
print(f"CM where k is {bestK}.")
cm = confusion_matrix(y,y_pred)
cm
```
