```python
class Customer:
    def __init__(self,customer_id,credit_limit,current_balance,orders):
        self.customer_id = customer_id
        self.credit_limit = credit_limit
        self.current_balance = current_balance
        self.orders = orders

    def __call__(self,x):
        self.credit_limit += self.credit_limit * x
        return self

    def addOrder(self,orderID):
        self.orders.append(orderID)
```

```python
x = Customer(10,10,100,[5,10,15])
print(x.credit_limit)
x = x(.10)
print(x.credit_limit)
```

```python
x.addOrder(20)
print(x.orders)
```
